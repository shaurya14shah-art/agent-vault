```python
word = input("Enter the word/number to be checked - ")


def is_palindrome(string=word):
    if (string[::-1] == string):
        return True
    else:
        return False
        
if word.isalnum():
    ans = is_palindrome(word)
    if ans:
        print(f"\"{word}\" is a palindrome.")
    else:
        print(f"\"{word}\" is not a palindrome.")
else:
    print("Invalid input. Please enter a string without spaces or special characters.")
```
