```python
password = ""
def validate_password(p = password):
    uppers = 0
    lowers = 0
    numbers = 0
    special = 0

    for i in  p:
        if i.isalpha():
            if i.isupper():
                uppers += 1
            elif not i.isupper():
                lowers += 1
        elif i.isdigit():
            numbers += 1
        elif not i.isalnum():
            special += 1

    if (len(p) >= 8):
        if (uppers >= 1):
            if (lowers >= 1):
                if(numbers >= 1):
                    print(f"Your password ({p}) successfully validated!")
                else:
                    print(f"there should be atleast 1 number in your password ({p})")
            else:
                print(f"there should be atleast 1 lowercase alphabet in your password ({p})")
        else:
            print(f"there should be atleast 1 uppercase alphabet in your password ({p})")
    else:
        print(f"there should be atleast 8 total characters in your password ({p})")

password = input("Enter your password - ")

validate_password(password)
```
